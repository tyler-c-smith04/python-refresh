```python
# Load Data
import pandas as pd

url = "https://github.com/nflverse/nfldata/raw/master/data/games.csv"
games = pd.read_csv(url)
```

```python
# Inspect data
print(games.shape)
games.head()
```

```python
# Learn about columns
games.info()
```

```python
# Select Columns
cols = ["season", "week", "away_team", "away_score", "home_team", "home_score", "result"]
g = games[cols]
g.head()
```

```python
# Filter Rows
season_2025 = g[g["season"] == 2025]
print(len(season_2025))
season_2025.head()
```

```python
# Filter two conditions at once
blowouts = g[(g["season"] >= 2020) & (g["result"] >= 30)]
blowouts
```

```python
# Sort Values
g.sort_values("result", ascending=False).head(10)
```

```python
# My team: filter g to games where my favorite team was home in 2025.
hou_home = g[(g["season"] == 2025) & (g["home_team"] == 'HOU')]
hou_home
```

```python
# Biggest road wins: sort to find the 5 games where the away team won by the most.
g.sort_values("result").head()
```

```python
# Coldest games: use the full games table, not g, to find the 5 coldest games by the temp column.
# Show only season, away_team, home_team, temp and stadium.
cols = ["season",  "away_team", "home_team", "temp", "stadium"]
games[cols].sort_values("temp").head()
```

```python
# How many home wins did the Texans have in 2025?
hou_home_wins = (hou_home["result"] > 0).sum()
print(hou_home_wins)
```
